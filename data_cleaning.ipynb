```python
writing_labels = {
    "1": "ChatGPT",
    "2": "Google Bard",
    "3": "Claude",
    "4": "Microsoft Bing AI",
    "5": "GrammarlyGo",
    "6": "Others",
    "7": "Never used any"
}

def convert_writing(value):
    if pd.isna(value):
        return value

    codes = str(value).split(",")
    names = []

    for code in codes:
        code = code.strip()
        names.append(writing_labels.get(code, code))

    return ", ".join(names)

df["genai_writing_tools"] = df["genai_writing_tools"].apply(convert_writing)
```

```python
def clean_writing_tool(value):
    if pd.isna(value) or str(value).strip() == "":
        return pd.NA

    value = " ".join(str(value).split()).lower()

    if value.startswith("grammarly"):
        return "Grammarly"

    tool_names = {
        "quillbot": "QuillBot",
        "perplexity": "Perplexity",
        "google translate": "Google Translate",
        "github copilot": "GitHub Copilot",
        "microsoft word's native editor suggestions": "Microsoft Word Editor Suggestions",
        "dall-e 2": "DALL·E 2",
        "lexis+ ai": "Lexis+ AI"
    }

    return tool_names.get(value, "Others")


df["other_genai_writing_tools"] = df["other_genai_writing_tools"].apply(clean_writing_tool)
```

```python
for i in df.index:
    main = df.loc[i, "genai_writing_tools"]
    other = df.loc[i, "other_genai_writing_tools"]

    if pd.isna(other):
        continue

    if pd.isna(main):
        df.loc[i, "genai_writing_tools"] = other

    elif "Others" in main:
        df.loc[i, "genai_writing_tools"] = main.replace("Others", other)

    elif other not in main.split(", "):
        df.loc[i, "genai_writing_tools"] = main + ", " + other
```

```python
df.drop(columns="other_genai_writing_tools", inplace=True)
```

```python
tool_keywords = {
    "chatgpt": "ChatGPT",
    "chat gpt": "ChatGPT",
    "cgatgpt": "ChatGPT",
    "ghatgpt": "ChatGPT",
    "gpt 4": "ChatGPT",
    "grammarly": "Grammarly",
    "claude": "Claude",
    "pictory": "Pictory AI",
    "bard": "Google Bard",
    "dalle": "DALL-E 2",
    "scite": "Scite",
    "openai api": "OpenAI API",
    "python api": "OpenAI API"
}
```

```python
def clean_paid_tools(value):
    if pd.isna(value) or str(value).strip() == "":
        return pd.NA

    text = str(value).lower()
    names = []

    for keyword, name in tool_keywords.items():
        if keyword in text and name not in names:
            names.append(name)

    if names:
        return ", ".join(names)
    else:
        return "Others"
```

```python
df["paid_genai_services"] = df["paid_genai_services"].apply(clean_paid_tools)
```
